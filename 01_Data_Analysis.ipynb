```python
crop_clean = crop.copy()

yield_clean = yield_data.copy()

sensor_clean = sensor.copy()

print("Backup datasets created successfully!")
```

```python
yield_clean["Irrigation"] = yield_clean["Irrigation"].fillna(
    yield_clean["Irrigation"].mode()[0]
)

yield_clean["Previous_Crop"] = yield_clean["Previous_Crop"].fillna(
    yield_clean["Previous_Crop"].mode()[0]
)

sensor_clean["irrigation_type"] = sensor_clean["irrigation_type"].fillna(
    sensor_clean["irrigation_type"].mode()[0]
)

sensor_clean["crop_disease_status"] = sensor_clean["crop_disease_status"].fillna(
    sensor_clean["crop_disease_status"].mode()[0]
)

print("Missing values handled successfully!")
```

```python
print("Yield Dataset Missing Values After Cleaning")
print(yield_clean.isnull().sum())
```

```python
print("Sensor Dataset Missing Values After Cleaning")
print(sensor_clean.isnull().sum())
```

```python
# ---------------- Crop Recommendation Dataset ----------------

crop_clean = crop_clean.rename(columns={
    "N": "nitrogen",
    "P": "phosphorus",
    "K": "potassium",
    "temperature": "temperature",
    "humidity": "humidity",
    "ph": "soil_ph",
    "rainfall": "rainfall",
    "label": "crop"
})

# ---------------- Crop Yield Dataset ----------------

yield_clean = yield_clean.rename(columns={
    "Crop": "crop",
    "Region": "region",
    "Soil_Type": "soil_type",
    "Soil_pH": "soil_ph",
    "Rainfall_mm": "rainfall",
    "Temperature_C": "temperature",
    "Humidity_pct": "humidity",
    "Fertilizer_Used_kg": "fertilizer_used_kg",
    "Pesticides_Used_kg": "pesticides_used_kg",
    "Planting_Density": "planting_density",
    "Yield_ton_per_ha": "yield_ton_per_ha",
    "Previous_Crop": "previous_crop",
    "Irrigation": "irrigation"
})

# ---------------- Smart Farming Dataset ----------------

sensor_clean = sensor_clean.rename(columns={
    "crop_type": "crop",
    "soil_moisture_%": "soil_moisture",
    "soil_pH": "soil_ph",
    "temperature_C": "temperature",
    "rainfall_mm": "rainfall",
    "humidity_%": "humidity",
    "yield_kg_per_hectare": "yield_kg_per_hectare",
    "NDVI_index": "ndvi",
    "crop_disease_status": "disease_status"
})

print("✅ Column names standardized successfully!")
```
